### 🧪 Project 2: "Run Node API as Pod"

Take your Node API and deploy it:

```yaml
apiVersion: v1
kind: Pod
metadata:
  name: node-api
  namespace: demo
  labels:
    app: node-api
    tier: backend
spec:
  containers:
    - name: node-api
      image: your-node-image
      ports:
        - containerPort: 3000
```

**Tasks:**
- `kubectl create namespace demo`
- `kubectl apply -f pod.yaml`
- `kubectl get pods -n demo`
- `kubectl describe pod node-api -n demo`
- `kubectl logs node-api -n demo`
- `kubectl get pods --show-labels`

### 🧠 Notes
- Pod = wrapper around container
- Pod is **ephemeral** (temporary)
- Namespace = virtual cluster inside cluster (isolation)
- Labels = key/value pairs for selecting/organizing
- Annotations = non-identifying metadata (tooling)

### ⚙️ Skills Gained
- Write Kubernetes YAML
- Deploy a pod in a namespace
- Inspect pods with describe/logs
- Use labels for filtering (`-l`, `--show-labels`)

## 🔧 Phase 9b: Specialized Workloads

### 📚 Learn
- **Job** — run a task to completion (not a long-running server)
- **CronJob** — schedule Jobs on a timer (like cron in Linux)
- **DaemonSet** — run exactly one pod on **every** node

### 🧪 Project 9b: "Ad-hoc & Scheduled Tasks"

**Job:** Run a one-off DB migration
```yaml
apiVersion: batch/v1
kind: Job
metadata:
  name: db-migrate
spec:
  template:
    spec:
      containers:
      - name: migrate
        image: your-migration-image
      restartPolicy: Never
```

**CronJob:** Schedule backup every hour
```yaml
apiVersion: batch/v1
kind: CronJob
metadata:
  name: backup
spec:
  schedule: "0 * * * *"
  jobTemplate:
    spec:
      template:
        spec:
          containers:
          - name: backup
            image: backup-image
          restartPolicy: Never
```

**DaemonSet:** Deploy a log collector on every node (demo)

**Tasks:**
- `kubectl apply -f job.yaml`
- `kubectl get jobs`
- `kubectl apply -f cronjob.yaml`
- `kubectl get cronjobs`
- `kubectl get daemonsets`

### 🧠 Notes
- **Job** → runs once, exits (migrations, batch processing)
- **CronJob** → runs on schedule (backups, reports, cleanup)
- **DaemonSet** → one-per-node (logging, monitoring, CNI)
- Deployment = long-lived stateless servers. These cover different patterns.

### ⚙️ Skills Gained
- Job vs CronJob vs DaemonSet
- When to use each workload type
- Scheduling tasks in K8s

## ❌ Out of Scope for This Bootcamp

These are **not covered** in the 3-day plan (not essential for the foundation):

| Resource / Topic | Why Omitted |
|-----------------|-------------|
| **Custom Resource Definitions (CRDs)** | You build apps, not K8s extensions |
| **Operators** | Advanced pattern — revisit after mastery |
| **Service Mesh (Istio/Linkerd)** | Adds complexity without need at this stage |
| **Helm Charts** | YAML templating — useful later, core K8s first |
| **Pod Security Policies / OPA** | Security hardening — learn basics first |
| **Cluster Autoscaler** | Cloud-specific, not relevant for kind |
| **Kubernetes Federation** | Multi-cluster management — far beyond scope |
| **Container runtimes (containerd vs CRI-O)** | Abstraction detail, not actionable yet |

> Focus on the 15 built-in resources listed above. Everything else can be learned *after* you're comfortable with those.
